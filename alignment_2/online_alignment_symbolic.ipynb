## 1. MIDI vs. Audio Online Alignment

### Input Representation:

- **Audio**
  - Derived from acoustic signals: 
    1. Chroma, 
    2. CQT + Spectral Flux, etc.

- **MIDI**
  - Derived from symbolic note events such as pitch, onset, offset, velocity: 
    1. Pitch + Onset, 
    2. Pitch Class + Onset,  
    3. Pianoroll, etc.

---


#### Advantages of MIDI
- No processing is required to decipher the musical note being played.
- Precise note onset and offset timing.
- Faster and cleaner feature extraction and matching --> lower latency.
- Robust to noise, timbre, and recording conditions.

#### Disadvantages of MIDI
- At best, provides approximated expressive acoustic information (timbre, articulation, dynamics).
- MIDI is most naturally suited to piano/keyboard-style/percussion note events;
  but not to instruments such as the violin or the human voice, 
  which would depend on the transcription of their audio signals into MIDI information.
- In comparison to audio recordings, fewer MIDI recordings of performances are available.

---

## 2. MIDI Features

The `MidiStream` in Matchmaker is responsible for sending the input MIDI data (either in windowed frames, or by note event) through a `Processor` that returns one feature vector.

Let's look at three processors in this part of the notebook:

- `PitchProcessor` — Merges the note-on events of an input frame into a single binary vector with the pitches that start in that frame. Depending on the selected pitch range, the processor outputs either an 88-bin vector (piano range) or a full 128-bin vector. This processor is used by the Pitch HMM and the Outer Product HMM methods.
- `ChordOnsetProcessor` — Processes incoming note events by setting the corresponding pitch indices in the feature vector to 1 whenever a note is played, and remains so until it has stopped playing. If multiple note events start to play within a 10 ms window, it produces a multi-hot vector representing all detected pitches. Depending on the selected pitch range, the processor outputs either an 88-bin vector (piano range) or a full 128-bin vector. This processor is used by the OLTW Arzt method.
- `ParticleFilterMidiProcessor` — Keeps track of the notes that are currently sounding, the notes that start in the current frame, and a loudness value derived from the MIDI velocities. This processor is used by the Particle Filter method.

````python
'''Uncomment the line below to manually set the method without using the dropdown.'''
#method = 'outerhmm' # 'pfkorz', 'arzt'

matchmaker = mm.Matchmaker(
    score_file=SNIPPET_SCORE_FILE,
    performance_file=MIDI_SNIPPET_FILE,
    input_type="midi",
    method=method,
)

sf = matchmaker.score_follower

if method == 'outerhmm':
    frame_skip = 3
elif method == 'pfkorz':
    frame_skip = 5
elif method == 'arzt':
    frame_skip = 3
# All processors use the piano range (88 keys)
pianoroll = np.zeros((88, 20))


def to_pianoroll_column(feature):
    """Piano-roll column (88 keys) of the features of one input frame"""
    if isinstance(feature, KorzeniowskiObservation):
        # The particle filter features include the MIDI pitches that are sounding
        column = np.zeros(88)
        column[feature.active_notes - 21] = 1
        return column
    return feature

pianoroll_cursor_end_point = pianoroll.shape[1] * 3 // 4
pianoroll_cursor_start_point = pianoroll.shape[1] * 2 // 5
def update_pianoroll(pianoroll, feature):
    pianoroll[:, :-1] = pianoroll[:, 1:]
    pianoroll[:, -1] = feature

    return pianoroll

fig_y = 8
fig_x = 16

fig_text = 45
post_stream = 15

fig, ax = plt.subplots(ncols=1, figsize=(fig_x, fig_y))
count = 1



def plot_features(pianoroll, pianoroll_cursor_end_point, pianoroll_cursor_start_point, fig_text, fig, ax, count):
    ax.clear()

    ax.imshow(pianoroll, aspect='auto', origin='lower')
    ax.set_title("Method: '{}'. Input Frame: {}".format(method, count))
    ax.set_xticks([])
            # add a vertical line at the cursor point
    ax.axvline(x=pianoroll_cursor_end_point+0.5, color='r', linestyle='--')
    ax.axvline(x=pianoroll_cursor_start_point-2.5, color='r', linestyle='--')
            # colour the background of the plot to the right of the cursor point till the end of the image in grey
    ax.axvspan(pianoroll_cursor_end_point+0.5, pianoroll.shape[1], color='purple')
    ax.axvspan(0, pianoroll_cursor_start_point-2.5, color='green')
    ax.set_xlim(0, pianoroll.shape[1])
            # add text to the right of the cursor point saying "MIDI Stream Processor"
    ax.text(pianoroll_cursor_end_point+1, fig_text, "<- MIDI Processor", fontsize=16, color='white')
    ax.text(1, fig_text, "Score Follower <--", fontsize=16)
    ax.set_xlabel("Input Time (frames)")
    ax.set_ylabel("Pitch")
    ax.grid(False, axis='y')

    clear_output(wait=True)
    display(fig)

flag=False
with matchmaker.stream:
    while not flag:
        features = sf.queue.get(timeout=QUEUE_TIMEOUT)
        if features is STREAM_END:
            print(f"End of stream reached with {count}. Plotting remaining frames...")
            for _ in range(post_stream):
                pianoroll = update_pianoroll(pianoroll, np.zeros_like(pianoroll.shape[0]))
                if count % frame_skip == 0 or count == 1:
                    plot_features(pianoroll, pianoroll_cursor_end_point, pianoroll_cursor_start_point, fig_text, fig, ax, count)
                count += 1
            break
        else:
            beat = sf(features[0], features[1])
            pianoroll = update_pianoroll(pianoroll, to_pianoroll_column(features[0]))

        if count % frame_skip == 0 or count == 1:
            plot_features(pianoroll, pianoroll_cursor_end_point, pianoroll_cursor_start_point, fig_text, fig, ax, count)                

        count += 1

ax.clear()
clear_output(wait=True)
plt.close()

````

### 3.2 Evaluating a run

To evaluate a run, we compare the alignment path of the score follower with the ground truth alignment in the `.match` file of the performance. The helper below does this with Matchmaker's evaluation utilities: `resolve_gt` reads the ground truth (the performance time of each score onset), and `evaluate_alignment` computes the evaluation metrics (see the evaluation section below). It also saves the alignment path and the ground truth (as TSV files) and the metrics (as a JSON file) in `results/`.

````python
def evaluate_run(mm, match_file, run_name):
    """Evaluate a finished Matchmaker run against the ground truth in a match file.

    Returns the evaluation metrics, the alignment path (performance time in seconds,
    score position in beats) and the ground truth (performance time in seconds, score
    position in beats).
    """
    # Row 0 of the alignment path is the performance time (seconds),
    # row 1 is the estimated score position (beats)
    perf_sec, score_beat = mm.score_follower.alignment_path.astype(float)

    # For MIDI input, the ground truth uses the onset times of the MIDI file that the
    # score follower received (instead of the ones in the match file)
    performance = mm.performance_file if mm.input_type == "midi" else None
    gt_perf, gt_score = resolve_gt(
        str(match_file), mm.score_part.note_array(), performance=performance
    )

    results = evaluate_alignment(score_beat, perf_sec, gt_score, gt_perf)
    # Real-time factor: processing time / performance duration
    results["rtf"] = round(mm.alignment_duration / (gt_perf.max() - gt_perf.min()), 4)
    if mm.input_type == "audio":
        # Average feature extraction and inference latencies per frame (ms)
        results.update(mm.get_latency_stats())

    # Save the alignment path, the ground truth and the metrics
    RESULTS_DIR.mkdir(exist_ok=True)
    for prefix, path in [("wp", (perf_sec, score_beat)), ("gt", (gt_perf, gt_score))]:
        np.savetxt(
            RESULTS_DIR / f"{prefix}_{run_name}.tsv",
            np.column_stack(path),
            delimiter="\t",
            header="perf_sec\tscore_beat",
            comments="",
        )
    with open(RESULTS_DIR / f"{run_name}.json", "w") as f:
        json.dump(results, f, indent=2)

    return results, perf_sec, score_beat, gt_perf, gt_score
````

## 5. MIDI score following

MIDI methods available in Matchmaker include:

- `"pthmm"`: pitch-based HMM score follower. This is the default for MIDI.
- `"outerhmm"`: outer-product HMM for symbolic input.
- `"arzt"` and `"dixon"`: event-based online time warping variants.
- `"pfkorz"`: particle filter for probabilistic score position and tempo estimation (Korzeniowski et al., 2014).

The default MIDI processor is the `PitchProcessor`, which merges the note events of an input frame into a single pitch observation.

### 5.1 PitchHMM

`pthmm` is the default MIDI method. It uses pitch observations and a probabilistic state model to estimate which score onset is currently active.

````python
print(f"Performance file: {MIDI_FILE.name}, with input mode: midi")
print(f"Running Matchmaker with score file ({SCORE_FILE.name}) and method 'pthmm'...")

midi_pthmm_mm = Matchmaker(
    score_file=SCORE_FILE,
    performance_file=MIDI_FILE,
    input_type="midi",
    method="pthmm",
)

fig, ax = plt.subplots(figsize=(10, 2))
midi_pthmm_positions = []
i = 0
for current_position in midi_pthmm_mm.run(verbose=False):
    ax.clear()
    ax.plot(current_position, 0, "ro")
    ax.annotate(f"Beat: {current_position:.2f}", xy=(current_position, 0), textcoords="offset points", xytext=(current_position, 2.5), ha='center', fontsize=12)
    ax.set_yticks([])
    ax.set_xlabel("Score Position (beats)")
    ax.set_title("Real-time Estimated Score Position")
    ax.set_xlim(0, score_positions.max()+1)
    clear_output(wait=True)
    display(fig)
    midi_pthmm_positions.append(float(current_position))
    i += 1

ax.clear()
plt.close()

midi_pthmm_path = midi_pthmm_mm.score_follower.alignment_path

print("MIDI observations:", len(midi_pthmm_positions))
print("Alignment path shape:", midi_pthmm_path.shape)
print("First 10 estimated beat positions:")
print(np.round(midi_pthmm_positions[:10], 3))

midi_pthmm_eval = evaluate_run(midi_pthmm_mm, MATCH_FILE, run_name="midi_pthmm")[0]

print("Evaluation result:")
print(json.dumps(midi_pthmm_eval, indent=2))
````

### 5.3 Particle Filter

A Particle Filter approaches the alignment task slightly differently. It tracks many hypothetical beat/tempo states (“particles”) as the performance input arrives, where each particle predicts where the next beat or score position should occur.
As new notes arrive, particles that better match the input receive higher weights, while poor matches are discarded or resampled into stronger positions.
Over time, the particle cloud converges on the most likely alignment between the performed audio and the reference score.

````python
pf_mm = None
pf_positions = []
pf_path = None
pf_eval = None

try:
    print(f"Performance file: {MIDI_FILE.name}, with input mode: midi")
    print(f"Running Matchmaker with score file ({SCORE_FILE.name}) and method 'pfkorz'...")

    pf_mm = Matchmaker(
        score_file=SCORE_FILE,
        performance_file=MIDI_FILE,
        input_type="midi",
        method="pfkorz",
    )

    fig, ax = plt.subplots(figsize=(10, 6))
    count = 0
    for current_position in pf_mm.run(verbose=False):
        if count % 20 == 0 or count == 1:
            ax.clear()
            pf = pf_mm.score_follower
            # Each particle has a score position (pf.x, in beats) and a tempo (pf.m, in log2 BPM)
            particle_tempi = pf.log2tempo_to_bpm(pf.m)
            tempo_mean = np.average(particle_tempi, weights=pf.weights)
            ax.plot(pf.x, particle_tempi, "r.", alpha=0.3, label="Particles")
            # plot the (weighted) mean tempo as a green cross along with its value
            ax.plot(current_position, tempo_mean, "gx", markersize=10, label="Estimated Tempo")
            ax.annotate(f"{tempo_mean:.2f} BPM", (current_position, tempo_mean), textcoords="offset points", xytext=(40,-5), ha='center', color='green')
            ax.plot(current_position, 0, "bx", markersize=10, label="Estimated Position")
            ax.annotate(f"Beat: {current_position:.2f}", (current_position, 0), textcoords="offset points", xytext=(0,10), ha='center', color='blue')
            ax.set_ylabel("Tempo (BPM)")
            ax.set_ylim(-10, 3 * pf_mm.tempo)
            ax.set_xlabel("Score Position (beats)")
            ax.set_title("Real-time Estimated Score Position")
            ax.set_xlim(0, score_positions.max() + 1)
            ax.legend()
            clear_output(wait=True)
            display(fig)
        pf_positions.append(float(current_position))
        count += 1

    ax.clear()
    plt.close()

    pf_path = pf_mm.score_follower.alignment_path

    print("Particle Filter observations:", len(pf_positions))
    print("Alignment path shape:", pf_path.shape)
    print("First 10 estimated beat positions:")
    print(np.round(pf_positions[:10], 3))

    pf_eval = evaluate_run(pf_mm, MATCH_FILE, run_name="pf")[0]

    print("Evaluation result:")
    print(json.dumps(pf_eval, indent=2))

except Exception as exc:
    print("Particle Filter demo skipped:")
    print(type(exc).__name__, exc)
````
